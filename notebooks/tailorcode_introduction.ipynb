# tAilorCode: From Customer Support AI to Autonomous Code Repair

This notebook is a lightweight public explanation of **tAilorCode**, my submission project for the [Google - The Gemma 4 Developer Agent Competition](https://www.kaggle.com/competitions/gemma-4-developer-agent).

The notebook does not load the 22 GB competition dataset. It explains the project and provides a compact structure for recording future evaluation results.

## The project story

My earlier KelceTS projects focused on grounded customer-support assistants: understand a message, use trusted information, follow rules, avoid hallucinations, and escalate when necessary.

`tAilorCode` transfers that way of thinking to software engineering. The customer message becomes a repository issue, the trusted knowledge becomes source code and tests, and the final response becomes a Git patch that must pass verification.

## 7. Train a baseline model

## 8. Evaluate the agent with metrics

The primary competition metric is **Resolution Rate**:

$$
\text{Resolution Rate} = \frac{\text{resolved tasks}}{\text{evaluated tasks}}
$$

A task is resolved only when the generated patch passes the required verification tests in a clean sandbox.
